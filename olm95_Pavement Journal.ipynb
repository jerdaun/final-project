Metric: Road Quality

This analysis defines Quality of Life based on the city's paving schedule, using two opposing interpretations of the data.

1. Primary Metric: Low Maintenance Required (Best Quality)
Streets with the lowest total completed paving activities are considered the "Best" because they are inherently well-built, require minimal intervention, and pose the lowest risk of potholes/damage to drivers.

2. Secondary Metric: High Maintenance Required (Worst Quality / Highest Priority)
Streets with the highest total completed paving activities are considered the "Worst" in terms of deterioration, but reflect the areas the city prioritizes most for maintenance.

The following code filters for completed work, aggregates activities by street, and identifies the two extreme groups:

```python
import pandas as pd

paving_file = 'f2eff713-dcdb-4ddd-8090-607d2d1f6eeb.csv'
paving_df = pd.read_csv(paving_file)

completed_paving = paving_df[paving_df['status'].str.contains('Completed', case=False, na=False)].copy()
completed_paving['street'] = completed_paving['street'].str.strip().str.upper()

paving_by_street = completed_paving.groupby('street').size().reset_index(name='Total_Completed_Activities')

best_streets = paving_by_street.sort_values(by='Total_Completed_Activities', ascending=True)
best_streets_low_activity = best_streets[best_streets['Total_Completed_Activities'] == 1].head(10).copy()

worst_streets_high_activity = paving_by_street.sort_values(by='Total_Completed_Activities', ascending=False).head(5).copy()

print("--- Top 10 Best Streets (Lowest Maintenance Required) ---")
print(best_streets_low_activity)

print("\n--- Top 5 Worst Streets (Highest Maintenance Required) ---")
print(worst_streets_high_activity)
```

These streets represent the highest quality infrastructure, requiring the least amount of city intervention.

Top 5 "Worst" Streets (Highest Maintenance Required)
These streets show the highest frequency of repair, indicating either poor initial quality or extreme traffic volume/wear.

The paving activity metric offers a two-sided view of Pittsburgh's infrastructure:
