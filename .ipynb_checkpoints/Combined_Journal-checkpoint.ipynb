```python
import pandas as pd
import numpy as np
```

Introduction:
    We are trying to find the best neighborhood in Pittsburgh by using 3 different metrics that are mostly independent from eachother to find a neighborhood that is the 'best' in a wide spread of metrics.

Metric:
    The metric will be the average of the rankings of the neighborhoods in each metric. The features we a measuring are paving schedule, fatal overdoses, and walkability/sidewalk to street ratio. 

Datasets:

    Paving Schedule: https://data.wprdc.org/dataset/paving-schedule
    Fatal Overdoses: https://data.wprdc.org/dataset/allegheny-county-fatal-accidental-overdoses/resource/1c59b26a-1684-4bfb-92f7-205b947530cf
    Walkability: https://data.wprdc.org/dataset/sidewalk-to-street-walkability-ratio

The Best Neighborhood:
    

```python
df1 = pd.read_csv("avg_by_neighborhood")
df2 = pd.read_csv("df")
#df3 = pd.read_csv("score_set_3.csv")

merged = df1.merge(df2, on="HOOD", how="outer")#.merge(df3, on="neighborhood", how="outer")

merged["final_score"] = merged[["score_out_of_10", "SCORE"]].mean(axis=1, skipna=True)
merged = merged.reset_index().sort_values(by="final_score", ascending=False)
merged.to_csv("Combined_Final_Score.csv", index=False)

merged
```

    The best neighborhood is a tie with Bedford Dwellings and Cedarhurst Manor both having 10 points.
